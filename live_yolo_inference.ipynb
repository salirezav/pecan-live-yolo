# Live YOLO segmentation — camera / video (OpenCV window)

Grab the **ELP global shutter USB camera** (or Spinnaker Blackfly / video file), run YOLO-seg on each frame, and overlay **Pecan** and **Kernel** masks in a plain OpenCV window (50% transparent, different colors).

All live-demo code lives in `scripts/live_yolo.py`. This notebook only imports that module. After you edit the `.py` file, re-run the import cell (it reloads the module), then the cells below.

## Python environment

Use the project `.venv` created by uv:

```powershell
uv sync
```

Then select the **`Pecan Live YOLO (.venv)`** kernel (Python 3.12, CUDA PyTorch, OpenCV, ultralytics).

Default camera backend is OpenCV DirectShow for the ELP UVC module (`Global Shutter Camera`). Spinnaker remains optional for Blackfly cameras — that needs a separate Python 3.10 env (`scripts/setup_spinnaker_env.ps1`) and `live.CAMERA_BACKEND = "spinnaker"`.

## Live-view controls

- The video window keeps the camera **aspect ratio** when you resize it.
- **Camera controls**: HQ Grab switches ELP MJPEG size (1280x720 ↔ 1920x1200). Spinnaker also gets Color Corr / Auto exposure-gain-WB sliders.
- Bottom-left stats: pecan count, cracked vs uncracked (kernel touching a pecan blob), and kernel-area / all-label-area.
- Play a file instead of the camera: `live.CAMERA_BACKEND = "video"` and set `live.VIDEO_PATH`, or `python scripts/live_yolo.py --video path.mp4`.
- **Drag** a rectangle on the video to set an ROI — only pecan/kernel masks **inside** that box are drawn. Press **`c`** to clear it.
- Press **`q`** or Esc to stop.

Desktop shortcut / `scripts/run_live_yolo.bat` runs the same `.py` file.

Run cells top to bottom. The live-view cell blocks until you press `q` or Esc.
